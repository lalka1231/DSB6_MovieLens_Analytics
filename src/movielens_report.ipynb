```python
# MovieLens: what can the first 1000 records tell us?
from movielens_analysis import Ratings, Tags, Movies, Links
data_path = '../datasets/ml-latest-small'
ratings = Ratings(data_path + '/ratings.csv')
ratings_movies = Ratings.Movies(ratings, data_path + '/movies.csv')
ratings_users = Ratings.Users(ratings)
tags = Tags(data_path + '/tags.csv')
movies = Movies(data_path + '/movies.csv')
links = Links(data_path + '/links.csv', data_path + '/movies.csv')
%timeit -n 1 -r 1 ratings_movies.dist_by_year()
```

```python
# When were the ratings created?
%timeit -n 1 -r 1 ratings_movies.dist_by_year()
ratings_movies.dist_by_year()
```

```python
# Which scores do viewers choose most often?
%timeit -n 1 -r 1 ratings_movies.dist_by_rating()
ratings_movies.dist_by_rating()
```

```python
# Which films attract the most ratings?
%timeit -n 1 -r 1 ratings_movies.top_by_num_of_ratings(5)
ratings_movies.top_by_num_of_ratings(5)
```

```python
# Which films have the best average score?
%timeit -n 1 -r 1 ratings_movies.top_by_ratings(5, 'average')
ratings_movies.top_by_ratings(5, 'average')
```

```python
# Which films divide viewers most strongly?
%timeit -n 1 -r 1 ratings_movies.top_controversial(5)
ratings_movies.top_controversial(5)
```
